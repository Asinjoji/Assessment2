```python
filepath='/content/student_performance.csv'
df=pd.read_csv(filepath)
df
```

```python
#Q11
print("First 5 records:", df.head(5))
print("Last 5 records.:", df.tail(5))
print("No of Rows:", df.shape[0])
print("no of Columns:", df.shape[1])
print("Column names:\n", df.columns.tolist())
print("Data types:\n", df.dtypes)
print("Statistical summary:\n", df.describe())
```

```python
df.sort_values(by='Final_Score',ascending=False)
print(df)
```

```python
#Q13
df['Average_Technical_Score']=df[['Python_Score','SQL_Score','Pandas_Score','NumPy_Score']].mean(axis=1)

df['Performance']=np.where(df['Average_Technical_Score']>=90,'Excellent',
np.where(df['Average_Technical_Score']>=75,'Good',
np.where(df['Average_Technical_Score']>=60,'Average','Needs Improvement')))

df
```

```python
#Q14
print('Number of students: ' ,df['Department'].value_counts())
print('Average Final Score: ' ,df.groupby('Department')['Final_Score'].mean())
print('Maximum Final Score: ' ,df.groupby('Department')['Final_Score'].max())
print('Minimum Final Score: ' ,df.groupby('Department')['Final_Score'].min())
```

```python
#Q15
final_scores = df['Final_Score'].to_numpy()


mean = np.nanmean(final_scores)
median = np.nanmedian(final_scores)
std = np.nanstd(final_scores)
p25 = np.nanpercentile(final_scores, 25)
p75 = np.nanpercentile(final_scores, 75)

print(f"Mean: {mean:.2f}")
print(f"Median: {median:.2f}")
print(f"Standard Deviation: {std:.2f}")
print(f"25th Percentile: {p25:.2f}")
print(f"75th Percentile: {p75:.2f}")


above_mean_students = df[df['Final_Score'] > mean]
print(f"\nNumber of students above the mean: {len(above_mean_students)}")
print(above_mean_students)
```

```python
#Q16
print("Missing values:\n", df.isnull().sum())
df.fillna({'Age': df.Age.mean()}, inplace=True)
df.fillna({'Attendance': df.Attendance.mean()}, inplace=True)
df.fillna({'Python_Score': df.Python_Score.mean()}, inplace=True)
df.fillna({'Pandas_Score': df.Pandas_Score.mean()}, inplace=True)
df.fillna({'NumPy_Score': df.NumPy_Score.mean()}, inplace=True)
df.fillna({'Final_Score':df.Final_Score.mean()}, inplace=True)
print(df)
```

```python
df['Average_Score']=df[['Python_Score','SQL_Score','Pandas_Score','NumPy_Score']].mean(axis=1)
df['Pass_Status']=np.where(df['Average_Score']>=50,'Pass','Fail')
df
```
